```python
import numpy
def W_ex(r):
    a = [.27, 28, .03, 5]
    return a[0] * numpy.exp(-a[1] * r**2)+a[2] * numpy.exp(-a[3] * r**2)
def W_in(r):
    b = [10, 30, .02, 3]
    return b[0]*r**2 * numpy.exp(-b[1] * r**2)+b[2] * numpy.exp(-b[3] * r**2)
def W(r):
    return numpy.where(r <= 1, W_ex(r) - W_in(r), 0)
```

```python
import numpy
x = numpy.arange(0,1,.01)
wex, win, w = W_ex(x), W_in(x), W(x)
```

```python
from bokeh.plotting import figure, show, output_notebook
output_notebook()
p = figure(x_axis_label='r', y_axis_label='value')
p.line(x, w, legend='W=Wex-Win', color='black')
p.line(x, wex, legend='Wex', color='red')
p.line(x, win, legend='Win', color='blue')
show(p)
```
