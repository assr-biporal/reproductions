```python
import brian2
from brian2 import mV, ms
import matplotlib.pyplot as plt
```

```python
brian2.start_scope()

tau = 10*ms
eqs = '''
dv/dt = (-g_m*(v-v_rest) + I_syn)/C_m : volt
g_m = 25*nS : siemens
C_m = 0.5*nF : farad
v_rest = -70*mV : volt
I_syn=0*mA : ampere
'''

G = brian2.NeuronGroup(1, eqs, threshold='v>-52*mV', reset='v=-59*mV', method='euler')

M = brian2.StateMonitor(G, 'v', record=0)
brian2.run(500*ms)
plt.plot(M.t/ms, M.v[0]/brian2.mV)
```
